```python
import numpy as np
import cv2
from ultralytics import YOLO
```

```python

# Load a pre-trained model (YOLO11n)
model = YOLO("yolo11n.pt")
```

```python

train_results = model.train(
  data="imagenet",
  epochs=100,
  imgsz=640,
  device="cpu"
)
```

```python
model.save("pretrained/imagenet.pt") 
```
